```python
# DAY 8: Evaluate relevance and summary quality

evaluation = []

for title in test_preferences:
    r = recommend_by_movie(title, 3)
    avg_similarity = round(r["similarity"].mean(), 3) if not r.empty else 0
    coverage = round(
        r["title"].isin(retrieval["title"]).mean(), 3
    ) if not r.empty else 0

    # Simple summary-quality score:
    summary_available = []
    for movie in r["title"]:
        s = review_summary(movie)
        summary_available.append(bool(s["pros"] or s["cons"]))
    summary_quality = round(np.mean(summary_available), 3) if summary_available else 0

    evaluation.append([title, avg_similarity, coverage, summary_quality])

evaluation_df = pd.DataFrame(evaluation, columns=[
    "Preference", "Average Similarity",
    "Review Retrieval Coverage", "Summary Quality"
])

display(evaluation_df)
```
