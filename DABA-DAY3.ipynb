**Task:** Connect recommendations with review sentiment.


```python
# DAY 3: Connect recommendations with review sentiment

def recommendation_with_sentiment(movie_title, n=5):
    rec = recommend_by_movie(movie_title, n)
    rows = []
    for _, row in rec.iterrows():
        s = review_summary(row["title"])
        rows.append([
            row["title"], row["genre"], row["rating"], row["similarity"],
            s["positive_reviews"], s["negative_reviews"], s["pros"], s["cons"]
        ])
    return pd.DataFrame(rows, columns=[
        "Movie","Genre","Rating","Similarity",
        "Positive Reviews","Negative Reviews","Pros","Cons"
    ])

display(recommendation_with_sentiment("Interstellar", 5))
```
