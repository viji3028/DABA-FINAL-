**Task:** Integrate recommender, sentiment and LLM.


```python
# DAY 13: Integrate recommender + sentiment + LLM prompt

def integrated_assistant(movie_title, user_preference="general"):
    rec = recommend_by_movie(movie_title, 5)
    outputs = []

    for _, row in rec.iterrows():
        s = review_summary(row["title"])

        prompt = improved_prompt.format(
            preference=user_preference,
            title=row["title"],
            genre=row["genre"],
            rating=row["rating"],
            similarity=round(row["similarity"], 3),
            pros=s["pros"],
            cons=s["cons"]
        )

        outputs.append({
            "movie": row["title"],
            "similarity": round(row["similarity"], 3),
            "pros": s["pros"],
            "cons": s["cons"],
            "llm_prompt": prompt
        })

    return outputs

integrated = integrated_assistant(
    "Interstellar",
    "science fiction and emotional movies"
)

for item in integrated[:3]:
    print("="*70)
    print("Movie:", item["movie"])
    print("Pros:", item["pros"])
    print("Cons:", item["cons"])
    print("\nLLM Prompt:\n", item["llm_prompt"])
```
